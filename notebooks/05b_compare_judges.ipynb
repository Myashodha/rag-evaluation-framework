```python
import pandas as pd

r = pd.read_csv("../reports/ragas_scores.csv")
d = pd.read_csv("../reports/deepeval_scores.csv")

r = r.rename(columns={
    "llm_context_precision_with_reference": "ragas_precision",
    "context_recall": "ragas_recall",
    "faithfulness": "ragas_faithfulness",
    "answer_relevancy": "ragas_relevancy",
})
m = r[["id", "ragas_precision", "ragas_recall", "ragas_faithfulness", "ragas_relevancy"]].merge(
    d[["id", "precision", "recall", "faithfulness", "relevancy"]], on="id")

for k in ["precision", "recall", "faithfulness", "relevancy"]:
    m[f"gap_{k}"] = (m[f"ragas_{k}"] - m[k]).abs()

print(m[[c for c in m.columns if c.startswith(("ragas_", "gap_")) or c in
         ["precision", "recall", "faithfulness", "relevancy"]]].mean(numeric_only=True).round(3))

# rows where the two judges disagree by more than 0.4 on any metric
gap_cols = [c for c in m.columns if c.startswith("gap_")]
disagree = m[(m[gap_cols] > 0.4).any(axis=1)]
display(disagree.round(2))
```
